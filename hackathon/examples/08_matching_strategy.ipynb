*"A1 is a dog, C3 is a car, D5 is a dog."* Each turn, you look at that memory and
pick the two best squares to flip.

## The game plan — in order of best to worst

Think about why this order makes sense:

1. **Free point!** If your memory already knows two squares with the *same* picture
   (like two dogs), flip those two. You know they match — easy point.

2. **Good guess.** If you don't have a sure pair, flip one card you *know* together
   with one card you've *never seen*. Worst case, you learn a new card. Best case,
   the new one matches!

3. **Explore.** If you don't know any cards yet, just flip two brand-new squares to
   start filling in your memory.

4. **Last resort.** If something weird happened and nothing above works, flip any two
   squares that haven't been matched yet, and let the referee sort it out.
